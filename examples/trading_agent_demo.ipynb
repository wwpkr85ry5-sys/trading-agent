# trading-agent demo

This notebook creates synthetic prices, trains the model, and prints a position series.

```python
import numpy as np
import pandas as pd

from trading_agent import Config, MLTradingAgent

rng = np.random.default_rng(42)
returns = rng.normal(0.0005, 0.01, 250)
prices = pd.Series(100.0 * np.exp(np.cumsum(returns)), index=pd.date_range('2020-01-01', periods=250, freq='D'))

agent = MLTradingAgent(Config())
position = agent.generate_position(prices)
print(position.head())
```
